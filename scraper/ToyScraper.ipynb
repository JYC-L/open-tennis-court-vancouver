```python
import requests
import pandas as pd
from datetime import datetime, timedelta

# Replace with your actual full URL
url = 'https://clubspark.ca/v0/VenueBooking/TBCHubRichmond/GetVenueSessions?resourceID=&startDate=2025-05-31&endDate=2025-06-02&roleId=&_=1748734993298'

# 'https://clubspark.ca/TBCHubRichmond/Booking/GetVenueSessions?resourceID=&startDate=2025-06-03&endDate=2025-06-03&roleId=&_=1748715748993'
headers = {
    'User-Agent': 'Mozilla/5.0'
}

def minutes_to_time(minutes):
    """Convert minutes since midnight to HH:MM format"""
    time = (datetime(2025, 6, 3) + timedelta(minutes=minutes)).time()
    return time.strftime('%H:%M')

response = requests.get(url, headers=headers)
if response.status_code == 200:
    data = response.json()
    session_list = []
    
    for resource in data['Resources']:
        court_name = resource.get('Name')
        for day in resource.get('Days', []):
            for session in day.get('Sessions', []):
                session_info = {
                    'Court': court_name,
                    'Date': day.get('Date', '').split('T')[0],
                    'Session Name': session.get('Name'),
                    'Start Time': minutes_to_time(session.get('StartTime')),
                    'End Time': minutes_to_time(session.get('EndTime')),
                    'Interval (min)': session.get('Interval'),
                    'Category': session.get('Category'),
                    'SubCategory': session.get('SubCategory'),
                    'Capacity': session.get('Capacity'),
                    'Restrictions': session.get('Restrictions', 'None'),
                    'Court Cost': session.get('CourtCost'),
                    'Guest Price': session.get('GuestPrice'),
                    'Member Price': session.get('MemberPrice')
                }
                session_list.append(session_info)
    
    # Convert to DataFrame
    df = pd.DataFrame(session_list)
    
    # Save to CSV
    df.to_csv('tennis_court_availability_2025-06-03.csv', index=False)
    print("✅ Data saved to 'tennis_court_availability_2025-06-03.csv'")

else:
    print(f"Failed to fetch data. Status code: {response.status_code}")
```
